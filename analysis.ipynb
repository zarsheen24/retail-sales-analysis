```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('Sample - Superstore.csv', encoding='latin1')

df.head()
df.info()

# Sales by category
sales_by_category = df.groupby('Category')['Sales'].sum()

df.groupby('Category')['Profit'].sum().plot(kind='bar')
plt.title('Profit by Category')
plt.show()

df['Order Date'] = pd.to_datetime(df['Order Date'])

sales_over_time = df.groupby('Order Date')['Sales'].sum()

sales_over_time.plot()
plt.title('Sales Over Time')
plt.show()

# Plot
sales_by_category.plot(kind='bar')
plt.title('Sales by Category')
plt.show()


## Key Insights

##Sales are highest in [insert category after checking]
## Certain categories generate more profit than others
## Sales increase during specific periods (e.g. end of year)
```
